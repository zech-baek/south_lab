```python
import importlib.util
import os
from pathlib import Path

parser_path = Path("parser") / "dcic_parser_v0.1.py"
spec = importlib.util.spec_from_file_location("dcic_parser_v0_1", parser_path)
dcic_parser_v0_1 = importlib.util.module_from_spec(spec)
spec.loader.exec_module(dcic_parser_v0_1)

log_parsing = dcic_parser_v0_1.parsing()
```

```python
dev = "sc85853"
keyword = "./parser/dcic_keyword_v0.1.yaml"
rev = "1p0"
log_path = "./log"
```

```python
# use directory list

list_log_file = [f for f in os.listdir(log_path) if os.path.isfile(os.path.join(log_path, f))]

for log_file in list_log_file:
    log_parsing.start_parsing(dump=f"./log/{log_file}", keyword=keyword, device=dev, revision=rev, vendor_keyword=True)
    print(f"\nprocessing done : {log_file}\n\n\n\n\n\n")
```
